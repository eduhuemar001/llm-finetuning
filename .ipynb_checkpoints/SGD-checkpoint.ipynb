# SGD Homework

Stochastic gradient descent of single-layer ANN with 2 input features, sigmoid activation and cross-entropy loss.

## Sigmoid activation function

$$
y = f(x; \mathbf{w}, b) = \frac{1}{1 + \exp(-\mathbf{w}^T \mathbf{x} - b)}
$$

## Cross-entropy loss

$$
J(y, y_{\text{ref}}) = -y_{\text{ref}} \log(y) - (1 - y_{\text{ref}}) \log(1 - y)
$$

$$
J(y, y_{\text{ref}}) =
\begin{cases}
-\log(y), & \text{if } y_{\text{ref}} = 1 \\
-\log(1 - y), & \text{if } y_{\text{ref}} = 0
\end{cases}
$$

## Gradient

$$
z = w_1 x_1 + w_2 x_2 + b
$$

$$
\nabla_{w,b}(J \circ f) =
\begin{cases}
-\dfrac{e^{-z}}{1 + e^{-z}} \cdot
\begin{pmatrix}
x_1 \\
x_2 \\
1
\end{pmatrix}, & \text{if } y_{\text{ref}} = 1 \\
\\
-\dfrac{1}{1 + e^{-z}} \cdot
\begin{pmatrix}
x_1 \\
x_2 \\
1
\end{pmatrix}, & \text{if } y_{\text{ref}} = 0
\end{cases}
$$


## SGD Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

# Gradient for y_ref = 1
def gradient_yref_1(x1, x2, w1, w2, b):
    z = w1 * x1 + w2 * x2 + b
    exp_neg_z = np.exp(-z)
    scalar = -exp_neg_z / (1 + exp_neg_z)
    return scalar * np.array([x1, x2, 1.0])

# Gradient for y_ref = 0
def gradient_yref_0(x1, x2, w1, w2, b):
    z = w1 * x1 + w2 * x2 + b
    scalar = -1 / (1 + np.exp(-z))
    return scalar * np.array([x1, x2, 1.0])

# Cross-entropy loss (single sample)
def compute_loss(x1, x2, w1, w2, b, y_ref):
    z = w1 * x1 + w2 * x2 + b
    y = 1 / (1 + np.exp(-z))
    if y_ref == 1:
        return -np.log(y + 1e-8)  # avoid log(0)
    else:
        return -np.log(1 - y + 1e-8)

# Data points
data = [
    ((0, 1), 1),
    ((1, 2), 1),
    ((1, 1), 1),
    ((0, 0), 0),
    ((1, 0), 0),
    ((2, 1), 0),
]

# Set random seed for reproducibility
np.random.seed(42)

# Learning rate and initial random parameters
eta = 0.05
w1 = np.random.randn()
w2 = np.random.randn()
b  = np.random.randn()

# Convergence threshold
epsilon = 1e-5
prev_loss = None

# Tracking for plotting
w1_list, w2_list, b_list = [], [], []

# SGD loop
for step in range(100):
    # Randomly select one data point
    (x1, x2), y_ref = data[np.random.randint(0, len(data))]

    # Compute loss before update
    loss = compute_loss(x1, x2, w1, w2, b, y_ref)

    # Check stopping condition
    if prev_loss is not None and abs(prev_loss - loss) < epsilon:
        print(f"Converged at step {step}, loss change < {epsilon}")
        break

    prev_loss = loss

    # Compute gradient based on y_ref
    if y_ref == 1:
        grad = gradient_yref_1(x1, x2, w1, w2, b)
    else:
        grad = gradient_yref_0(x1, x2, w1, w2, b)

    # Update weights and bias
    w1 -= eta * grad[0]
    w2 -= eta * grad[1]
    b  -= eta * grad[2]

    # Track parameters for plotting
    w1_list.append(w1)
    w2_list.append(w2)
    b_list.append(b)

    # Print progress
    print(f"Step {step:3d} | Sample: ({x1},{x2}) y_ref={y_ref} | "
          f"w1={w1:.4f}, w2={w2:.4f}, b={b:.4f} | "
          f"grad=({grad[0]:.4f}, {grad[1]:.4f}, {grad[2]:.4f}) | "
          f"loss={loss:.6f}")
```

```python
# Plot parameter evolution
plt.figure(figsize=(10, 6))
plt.plot(w1_list, label='w1', color='blue')
plt.plot(w2_list, label='w2', color='green')
plt.plot(b_list, label='b', color='red')
plt.xlabel("Iteration")
plt.ylabel("Parameter Value")
plt.title("SGD Parameter Evolution")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
total_loss = 0
for (x1, x2), y_ref in data:
    total_loss += compute_loss(x1, x2, w1, w2, b, y_ref)
avg_loss = total_loss / len(data)
print(f"\nAverage loss for all data points: {avg_loss:.6f}")
```
