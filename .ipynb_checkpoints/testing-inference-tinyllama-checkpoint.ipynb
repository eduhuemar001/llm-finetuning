```python
import json
import time
import psutil
import pandas as pd
from sklearn.metrics import classification_report
from huggingface_hub import hf_hub_download
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    AutoModelForSequenceClassification,
    pipeline
)
import torch
```

```python
# ========== Load and Prepare Dataset ==========
json_path = hf_hub_download(
    repo_id="eduhuemar001/tinyllama-german-dataset-sentiment-temp-test",
    filename="test_dataset.json",
    repo_type="dataset"
)

with open(json_path, "r", encoding="utf-8") as f:
    data = [json.loads(line) for line in f if line.strip()]

df = pd.DataFrame(data)
df = df[["review_text", "sentiment"]]
df = df.dropna(subset=["review_text", "sentiment"])
df["review_text"] = df["review_text"].astype(str).str.strip()
df["sentiment"] = df["sentiment"].astype(str).str.lower().str.strip()
df = df[df["sentiment"].isin(["positive", "neutral", "negative"])]

# Sample data
df = df[:1000]

texts = df["review_text"].tolist()
labels = df["sentiment"].tolist()
valid_outputs = ["positive", "neutral", "negative"]

instruction = (
    "### Instruction:\n"
    "Klassifiziere die Stimmung der folgenden Bewertung als 'positive', 'neutral' oder 'negative'.\n\n"
    "### Bewertung:\n"
)
answer_prefix = "\n\n### Antwort:\n"

# ========== Evaluation Helpers ==========
def print_eval_stats(name, y_true, y_pred, valid_count, total, wallclock, cpu_time):
    print(f"\n--- {name} ---")
    print(f"Wallclock-Zeit: {wallclock:.2f}s")
    print(f"CPU-Zeit: {cpu_time:.2f}s")
    print(f"Gültige Modellantworten: {valid_count} von {total} ({100 * valid_count / total:.2f}%)")
    print("\nKlassifikationsbericht (nur gültige Vorhersagen):")
    print(classification_report(y_true, y_pred, digits=3))


# ========== TinyLlama Inference ==========
def test_tinyllama_lora():
    print("\n[1/2] Lade TinyLlama-Modell...")
    load_start = time.time()
    cpu_start = time.process_time()

    model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
    lora_repo_id  = "eduhuemar001/tinyllama-german-checkpoints-sentiment"
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    base_model = AutoModelForCausalLM.from_pretrained(model_name)
    lora_model = PeftModel.from_pretrained(base_model, lora_repo_id)
    lora_model = lora_model.to(device)
    lora_model.eval()


    load_end = time.time()
    cpu_end = time.process_time()
    print(f"→ Ladezeit TinyLlama: {load_end - load_start:.2f}s (Wallclock), {cpu_end - cpu_start:.2f}s (CPU)")

    pred_labels = []
    true_labels = []
    valid_count = 0

    print("\nStarte Inferenz mit TinyLlama...")
    inf_start = time.time()
    cpu_inf_start = time.process_time()

    for i, row in df_balanced.iterrows():
        prompt = instruction + row["review_text"] + answer_prefix
        inputs = tokenizer(prompt, return_tensors="pt")

        with torch.no_grad():
            output = lora_model.generate(**inputs, max_new_tokens=2)

        decoded = tokenizer.decode(output[0], skip_special_tokens=True)
        if "### Antwort:" in decoded:
            answer = decoded.split("### Antwort:")[-1].strip().lower().split()[0]
        else:
            answer = decoded.strip().lower().split()[0]

        pred_labels.append(answer)
        true_labels.append(row["sentiment"])
        if answer in valid_outputs:
            valid_count += 1

    inf_end = time.time()
    cpu_inf_end = time.process_time()

    print_eval_stats(
        name="TinyLlama (LoRA, CPU)",
        y_true=true_labels,
        y_pred=pred_labels,
        valid_count=valid_count,
        total=len(df_balanced),
        wallclock=inf_end - inf_start + (load_end - load_start),
        cpu_time=cpu_inf_end - cpu_inf_start + (cpu_end - cpu_start)
    )


# ========== BERT Inference ==========
def test_bert():
    print("\n[2/2] Lade BERT-Modell...")
    load_start = time.time()
    cpu_start = time.process_time()

    model_name = "oliverguhr/german-sentiment-bert"
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSequenceClassification.from_pretrained(model_name)
    classifier = pipeline("sentiment-analysis", model=model, tokenizer=tokenizer, device=-1)

    load_end = time.time()
    cpu_end = time.process_time()
    print(f"→ Ladezeit BERT: {load_end - load_start:.2f}s (Wallclock), {cpu_end - cpu_start:.2f}s (CPU)")

    pred_labels = []
    true_labels = []
    valid_count = 0

    print("\nStarte Inferenz mit BERT...")
    inf_start = time.time()
    cpu_inf_start = time.process_time()

    results = classifier(texts, truncation=True)

    for i, res in enumerate(results):
        label = res["label"].lower()
        true = labels[i]
        pred_labels.append(label)
        true_labels.append(true)
        if label in valid_outputs:
            valid_count += 1

    inf_end = time.time()
    cpu_inf_end = time.process_time()

    print_eval_stats(
        name="BERT (CPU)",
        y_true=true_labels,
        y_pred=pred_labels,
        valid_count=valid_count,
        total=len(df_balanced),
        wallclock=inf_end - inf_start + (load_end - load_start),
        cpu_time=cpu_inf_end - cpu_inf_start + (cpu_end - cpu_start)
    )


# ========== Run Tests ==========
if __name__ == "__main__":
    test_tinyllama_lora()
    test_bert()
```
